```python
# Tuple Programs - Student Management System

# 1. Create Tuple
student = (101, "Rajeshree", "Computer", 85)

print("Student Tuple:", student)

# 2. Indexing
print("Roll Number:", student[0])
print("Student Name:", student[1])

# 3. Packing
details = (102, "Sneha", "Science")
print("Packed Tuple:", details)

# 4. Unpacking
roll, name, department = details

print("Roll:", roll)
print("Name:", name)
print("Department:", department)

# 5. Count Method
marks = (80, 90, 85, 90, 75)

print("Count of 90:", marks.count(90))

# 6. Index Method
print("Index of 85:", marks.index(85))

# 7. Slicing
print("First Three Marks:", marks[:3])

# 8. Immutability
# Tuple cannot be modified directly

try:
    student[1] = "Amit"
except TypeError:
    print("Tuple is immutable. Values cannot be changed.")
```
